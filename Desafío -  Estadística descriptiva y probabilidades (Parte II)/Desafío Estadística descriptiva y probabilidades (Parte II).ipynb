```python
#j. pertenecer a Asia y clasificar a segunda ronda
prob_asia = ((df['continent'] == 'asia') & (df['clasificado'] == 1)).mean()
prob_asia
```

```python
#l. pertenecer a Norteamérica y clasificar a segunda ronda
prob_norteamerica = ((df['continent'] == 'northamerica') & (df['clasificado'] == 1)).mean()
prob_norteamerica
```
